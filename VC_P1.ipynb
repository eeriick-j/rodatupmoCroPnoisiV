```python
# Crear una imagen negra de 800x800 píxeles
tablero = np.zeros((800, 800), dtype=np.uint8)

# Tamaño de cada casilla
tam_casilla = 100

# Crear el tablero de ajedrez
for fila in range(8): 
    for columna in range(8):
        if (fila + columna) % 2 == 0:
            tablero[fila*tam_casilla:(fila+1)*tam_casilla,
                    columna*tam_casilla:(columna+1)*tam_casilla] = 255

# Mostrar la imagen
plt.imshow(tablero, cmap='gray', vmin=0, vmax=255)
plt.show()
```

Separa los tres planos de la imagen captada por la webcam y muestra en horizontal

```python
vid = cv2.VideoCapture(0)
  
while(True):      
    # fotograma a fotograma
    ret, frame = vid.read()

    if ret:
        #Separamos canales
        b = frame[:,:,0]
        g = frame[:,:,1]
        r = frame[:,:,2]
        #Dimensiones imagen de entrada
        h, w, c = frame.shape

        #Concateamos en horizontal los tres planos del fotograma
        collage = np.hstack((r, g, b))
    
        # Muestra fotograma redimensionando a la mitad para que quepa en pantalla
        cv2.imshow('RGB', cv2.resize(collage, (int(w*1.5),int(h/2)),cv2.INTER_NEAREST))
    
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

Evento de ratón para mostrar el valor RGB del píxel

```python
# Función que trata eventos de ratón 
def mouse_events(event, x, y, flags, params):
    global px, py, R, G, B
    px = -1
    # Movimiento del puntero     
    if event==cv2.EVENT_MOUSEMOVE:
        #Valores del píxel, desde la webcam no es BGR
        B = frame[y, x, 0]
        G = frame[y, x, 1]
        R = frame[y, x, 2]  
        # Almacena valores del evento
        px = x
        py = y

     
  
```

```python
vid = cv2.VideoCapture(0)

# Tipografía para mostrar texto
font = cv2.FONT_HERSHEY_SIMPLEX
  
lanzado = 0 #bandera para primer fotograma
px = -1
while(True):      
    # fotograma a fotograma
    ret, frame = vid.read()

    if ret: 
        # Activa el manejador en el primer fotograma
        if lanzado == 0:
            # Muestra fotograma
            cv2.imshow('Cam', frame)   
            # Define el nombre del manejador del evento
            cv2.setMouseCallback('Cam', mouse_events)
            lanzado = 1
            
        # Muestra valores RGB intentando centrar en el puntero
        if px > -1:
            cv2.putText(frame, '{}'.format(R), (px-45,py-5), font, 0.5, (0, 0, 255), 1)
            cv2.putText(frame, '     {}'.format(G), (px-54,py-5), font, 0.5, (0, 255, 0), 1)
            cv2.putText(frame, '         {}'.format(B), (px-54,py-5), font, 0.5, (255, 0, 0), 1)
            
        cv2.imshow('Cam', frame)   
        
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

Transformando la imagen capturada en una colección de círculos, cuyo radio depende de la intensidad promedio de la zona

```python
vid = cv2.VideoCapture(0)

# Se define el número de celdas
ncells = 10
off = int(ncells/2)
  
while(True):      
    # fotograma a fotograma
    ret, frame = vid.read()

    if ret:
        #Dimensiones originales
        h, w, c = frame.shape
        #Redimensiono teniendo en cuenta ncells
        down_frame = cv2.resize(frame, (int(w/ncells),int(h/ncells)),cv2.INTER_NEAREST)
        #Dimensiones reducidas
        h2, w2, c2 = down_frame.shape

        #Separamos canales
        r = down_frame[:,:,0]
        g = down_frame[:,:,1]
        b = down_frame[:,:,2]

        # Creamos imagen negra
        gris_up_frame = np.zeros((h2*ncells,w2*ncells,1), dtype = np.uint8)

        # recorre valores de la imagen reducida
        for y in range(0,h2):
            for x in range(0,w2):
                # La suma de los valores RGB define el radio del círculo
                rad = int ( (r[y,x] + g[y,x] + b[y,x])/(ncells*3*2))
                cv2.circle(gris_up_frame,(x*ncells+off, y*ncells + off), rad, (255,255,255), -1)

    
        # Muestra fotograma resultante
        cv2.imshow('Cam', gris_up_frame)
    
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

vid = cv2.VideoCapture(0)

# Dimensiones de la cámara reducidas para fluidez
w = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH) / 2)
h = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT) / 2)
vid.set(cv2.CAP_PROP_FRAME_WIDTH, w)
vid.set(cv2.CAP_PROP_FRAME_HEIGHT, h)

# Creamos la imagen conjunta de 2x2 para el estilo cuadrícula Pop Art
collage = np.zeros((h*2, w*2, 3), dtype=np.uint8)
tl = collage[0:h, 0:w]         # Superior Izquierda
tr = collage[0:h, w:w+w]     # Superior Derecha
bl = collage[h:h+h, 0:w]       # Inferior Izquierda
br = collage[h:h+h, w:w+w]     # Inferior Derecha

while True:
    ret, frame = vid.read()
    if not ret:
        break

    # Redimensionamos
    frame = cv2.resize(frame, (w, h), interpolation=cv2.INTER_NEAREST)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Aplicamos un filtro de umbral para conseguir el contraste de cómic
    _, thresh = cv2.threshold(gray, 110, 255, cv2.THRESH_BINARY)
    
    # Convertimos la máscara binaria a 3 canales (0 a 1 en lugar de 0 a 255 para multiplicar bien)
    mask = thresh.astype(np.float32) / 255.0
    mask_3ch = cv2.merge([mask, mask, mask])

    # Asignamos combinaciones de colores vibrantes a cada panel multiplicando correctamente
    tl[:] = (mask_3ch * np.array([0, 255, 255], dtype=np.float32)).astype(np.uint8)   # Amarillo
    tr[:] = (mask_3ch * np.array([255, 0, 255], dtype=np.float32)).astype(np.uint8)   # Magenta
    bl[:] = (mask_3ch * np.array([255, 255, 0], dtype=np.float32)).astype(np.uint8)   # Cian
    br[:] = (mask_3ch * np.array([0, 255, 100], dtype=np.float32)).astype(np.uint8)   # Verde

    # Mostramos el resultado estilo galería Pop Art
    cv2.imshow('Pop Art - Era Digital', collage)

    # Salir pulsando la tecla ESC (27)
    if cv2.waitKey(20) == 27:
        break

vid.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

ancho = 1000
alto = 700

vid = cv2.VideoCapture(0)

while True:
    ret, frame = vid.read()
    
    if ret:
        # 1. Procesamiento que ya tenías para la cámara
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)        
        min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(gray)

        cv2.circle(frame, min_loc, 10, (255, 0, 255), -1)
        cv2.circle(frame, max_loc, 10, (0, 255, 255), -1)

        # 2. Crear la imagen base del panel en cada iteración
        image = np.full((alto, ancho, 3), dtype=np.uint8, fill_value=255)

        # Rectángulo base y zonas de colores
        cv2.rectangle(image, (0, 0), (ancho, alto), (0, 0, 0), 5)
        cv2.rectangle(image, (0, 0), (ancho, alto // 5), (255, 0, 0), -1)
        cv2.rectangle(image, (0, 0), (ancho, alto // 5), (0, 0, 0), 5)
        cv2.rectangle(image, (0, alto), (ancho // 2, alto - alto // 5), (0, 255, 0), -1)
        cv2.rectangle(image, (0, alto), (ancho // 2, alto - alto // 5), (0, 0, 0), 5)
        cv2.rectangle(image, (ancho - ancho // 5 + 5, alto - alto // 3), (ancho - 5, alto - 5), (255, 255, 0), -1)

        # 3. Coordenadas de la Zona azul
        x1 = ancho // 5 + 5
        y1 = alto // 5 + 5
        x2 = ancho - ancho // 5 - 5
        y2 = alto - alto // 5 - 5

        ancho_azul = x2 - x1
        alto_azul = y2 - y1

        # 4. Redimensionar el frame procesado al tamaño del rectángulo azul e insertarlo
        frame_redimensionado = cv2.resize(frame, (ancho_azul, alto_azul))
        image[y1:y2, x1:x2] = frame_redimensionado

        # Borde para la zona azul
        cv2.rectangle(image, (x1, y1), (x2, y2), (0, 0, 0), 2)

        # 5. Líneas divisorias finales
        cv2.line(image, (ancho // 5, 0), (ancho // 5, alto), (0, 0, 0), 5)
        cv2.line(image, (ancho - ancho // 5, 0), (ancho - ancho // 5, alto), (0, 0, 0), 5)
        cv2.line(image, (ancho // 2, alto - alto // 5), (ancho - ancho // 5, alto - alto // 5), (0, 0, 0), 5)
        cv2.line(image, (ancho - ancho // 5, alto - alto // 3), (ancho, alto - alto // 3), (0, 0, 0), 5)

        # Mostrar el resultado final compuesto
        cv2.imshow('Cam', image)
    
    if cv2.waitKey(20) == 27:
        break

vid.release()
cv2.destroyAllWindows()
```
